```python
import pandas as pd
import ast
import numpy as np
import re

def limpar_e_converter(texto):
    # 1. Remove 'np.float64(' mas mantém o número
    # O \d+\.?\d* captura o número (inteiro ou decimal)
    texto_limpo = re.sub(r'np\.float64\((.*?)\)', r'\1', texto)
    
    # 2. Agora a string é uma tupla Python válida: ({1: 0.66, ...}, 1.24)
    return ast.literal_eval(texto_limpo)
```

```python
dbpedia = pd.read_csv('../results/dbpedia/lql.csv')
```

```python
dbpedia = pd.read_csv('../results/dbpedia/lql.csv')
dbpedia['hae'] = dbpedia['hae'].apply(limpar_e_converter)

hae_split = pd.DataFrame(dbpedia['hae'].tolist(), index=dbpedia.index)
hae_split.columns = ['hae_level', 'hae_total']

dbpedia = pd.concat([dbpedia, hae_split], axis=1).drop(columns=['hae'])

dbpedia.head()
```
